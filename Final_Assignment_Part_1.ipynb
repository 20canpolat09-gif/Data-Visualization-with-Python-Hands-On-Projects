# Final Assignment: Part 1 - Create Visualizations using Matplotlib, Seaborn & Folium
This notebook contains the complete code and visualizations for analyzing historical automobile sales data during recession and non-recession periods.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium

# Load Data
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv'
df = pd.read_csv(URL)
print('Data downloaded and loaded successfully!')
df.head()
```

### Task 1.1: Line chart showing automobile sales fluctuation year-to-year

```python
df_line = df.groupby('Year')['Automobile_Sales'].mean().reset_index()
plt.figure(figsize=(10, 6))
plt.plot(df_line['Year'], df_line['Automobile_Sales'], marker='o')
plt.title('Automobile Sales Fluctuation Over Time')
plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.grid(True)
plt.show()
```

### Task 1.2: Line chart showing Advertising Expenditure vs Sales during Non-Recession periods

```python
df_non_rec = df[df['Recession'] == 0]
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_non_rec, x='Advertising_Expenditure', y='Automobile_Sales', ci=None)
plt.title('Advertising Expenditure vs Automobile Sales (Non-Recession)')
plt.xlabel('Advertising Expenditure ($)')
plt.ylabel('Automobile Sales')
plt.show()
```

### Task 1.3: Bar chart comparing sales per vehicle type in Recession vs Non-Recession

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Vehicle_Type', y='Automobile_Sales', hue='Recession')
plt.title('Average Vehicle Sales by Type: Recession vs Non-Recession')
plt.xlabel('Vehicle Type')
plt.ylabel('Average Sales')
plt.legend(title='Recession')
plt.show()
```

### Task 1.4: Subplots comparing GDP during Recession and Non-Recession periods

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)
sns.lineplot(data=df[df['Recession'] == 1], x='Year', y='GDP', ax=axes[0], color='red', marker='o')
axes[0].set_title('GDP Variations During Recession Periods')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('GDP')

sns.lineplot(data=df[df['Recession'] == 0], x='Year', y='GDP', ax=axes[1], color='green', marker='o')
axes[1].set_title('GDP Variations During Non-Recession Periods')
axes[1].set_xlabel('Year')
plt.show()
```

### Task 1.5: Bubble plot displaying impact of seasonality on Automobile Sales

```python
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x='Month', y='Automobile_Sales', size='Seasonality_Weight', sizes=(20, 500), hue='Seasonality_Weight', alpha=0.7)
plt.title('Impact of Seasonality on Automobile Sales')
plt.xlabel('Month')
plt.ylabel('Automobile Sales')
plt.show()
```

### Task 1.6: Scatter plot for correlation between average vehicle price and sales volume during recessions

```python
df_rec = df[df['Recession'] == 1]
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_rec, x='Price', y='Automobile_Sales')
plt.title('Vehicle Price vs Sales Volume During Recessions')
plt.xlabel('Average Vehicle Price ($)')
plt.ylabel('Sales Volume')
plt.show()
```

### Task 1.7: Pie chart displaying portion of advertising expenditure during Recession vs Non-Recession

```python
ad_rec = df.groupby('Recession')['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 8))
plt.pie(ad_rec, labels=['Non-Recession', 'Recession'], autopct='%1.1f%%', startangle=90, colors=['#66b3ff','#ff9999'])
plt.title('Total Advertising Expenditure: Recession vs Non-Recession')
plt.show()
```

### Task 1.8: Pie chart displaying total Advertisement expenditure for each vehicle type during Recession

```python
df_rec_ad = df_rec.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 8))
plt.pie(df_rec_ad, labels=df_rec_ad.index, autopct='%1.1f%%', startangle=90)
plt.title('Advertising Expenditure Share by Vehicle Type During Recessions')
plt.show()
```

### Task 1.9: Line chart analyzing effect of unemployment rate on vehicle type and sales during Recession

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_rec, x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type', ci=None, marker='o')
plt.title('Effect of Unemployment Rate on Vehicle Sales by Type (Recession)')
plt.xlabel('Unemployment Rate')
plt.ylabel('Automobile Sales')
plt.legend(title='Vehicle Type')
plt.show()
```
